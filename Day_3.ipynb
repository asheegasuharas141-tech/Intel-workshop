```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier, plot_tree

# 1. Reload raw dataset cleanly to wipe out the method object in Glucose
diabetes_path = "/content/diabetes.csv"
diabetes_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
    "Outcome",
]
diabetes = pd.read_csv(diabetes_path, names=diabetes_columns).drop_duplicates()

diabetes_input_columns = [
    "Pregnancies",
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
    "DiabetesPedigreeFunction",
    "Age",
]
diabetes_target_column = "Outcome"

# 2. Replace impossible zeros with NaN
invalid_zero_columns = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI",
]
diabetes[invalid_zero_columns] = diabetes[invalid_zero_columns].replace(0, np.nan)

# 3. Impute every column that has NaNs using actual called values (.median() / .mean())
for col in diabetes_input_columns:
    if diabetes[col].isnull().sum() > 0:
        skewness = diabetes[col].skew()
        if skewness < -0.5 or skewness > 0.5:
            diabetes[col] = diabetes[col].fillna(diabetes[col].median())
        else:
            diabetes[col] = diabetes[col].fillna(diabetes[col].mean())

print("Remaining missing values:\n", diabetes.isnull().sum())

# 4. Split, fit, and evaluate
x_diabetes = diabetes[diabetes_input_columns]
y_diabetes = diabetes[diabetes_target_column]

x_train_diabetes, x_test_diabetes, y_train_diabetes, y_test_diabetes = train_test_split(
    x_diabetes,
    y_diabetes,
    test_size=0.20,
    random_state=42,
    stratify=y_diabetes,
)

decision_tree_model = DecisionTreeClassifier(
    max_depth=4,
    class_weight="balanced",
    random_state=42,
)
decision_tree_model.fit(x_train_diabetes, y_train_diabetes)
print("\nDecision tree model trained successfully")

diabetes_predictions = decision_tree_model.predict(x_test_diabetes)
print("Accuracy:", accuracy_score(y_test_diabetes, diabetes_predictions))
print("\nConfusion Matrix:\n", confusion_matrix(y_test_diabetes, diabetes_predictions))
print("\nClassification Report:\n", classification_report(y_test_diabetes, diabetes_predictions))

# 5. Plot tree cleanly
plt.figure(figsize=(18, 8))
plot_tree(
    decision_tree_model,
    feature_names=diabetes_input_columns,
    class_names=["No Diabetes", "Diabetes"],
    filled=True,
    rounded=True,
)
plt.title("Decision Tree for Diabetes prediction")
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# 1. Train Random Forest on the cleaned train/test split from cell [114]
rf_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=6,
    class_weight="balanced",
    random_state=42,
)
rf_model.fit(x_train_diabetes, y_train_diabetes)

# 2. Evaluate on the test set
rf_predictions = rf_model.predict(x_test_diabetes)
print("Random Forest Accuracy:", accuracy_score(y_test_diabetes, rf_predictions))
print("\nConfusion Matrix:\n", confusion_matrix(y_test_diabetes, rf_predictions))
print("\nClassification Report:\n", classification_report(y_test_diabetes, rf_predictions))

# 3. Feature importance ranking
importances = pd.Series(
    rf_model.feature_importances_, index=x_train_diabetes.columns
).sort_values(ascending=False)
print("\nFeature Importances:\n", importances)

plt.figure(figsize=(10, 5))
importances.sort_values().plot(kind="barh")
plt.title("Random Forest Feature Importances (Diabetes)")
plt.xlabel("Importance")
plt.tight_layout()
plt.show()

# 4. Predict on `unknown_prediction` from cells [122]-[124]
unknown_aligned = unknown_prediction[x_train_diabetes.columns]
unknown_pred = rf_model.predict(unknown_aligned)[0]
unknown_prob = rf_model.predict_proba(unknown_aligned)[0]

label = "Diabetes (1)" if unknown_pred == 1 else "No Diabetes (0)"
print(f"\nPrediction for unknown_prediction: {label}")
print(
    f"Probabilities -> No Diabetes: {unknown_prob[0]:.2%}, Diabetes: {unknown_prob[1]:.2%}"
)
```
